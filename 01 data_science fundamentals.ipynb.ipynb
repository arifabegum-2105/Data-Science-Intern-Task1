```python
# Dataset
import numpy as np
import pandas as pd

np.random.seed(42)  # reproducibility

python_marks = [80, 75, 90, 60, 85, 70, 95, 65, 78, 88]
stats_marks  = [72, 70, 85, 55, 80, 68, 90, 60, 74, 82]

df = pd.DataFrame({"Python": python_marks, "Statistics": stats_marks})
df
```

```python
# Mean, median, mode
data = python_marks

mean = sum(data) / len(data)

sorted_data = sorted(data)
n = len(sorted_data)
if n % 2 == 0:
    median = (sorted_data[n//2 - 1] + sorted_data[n//2]) / 2
else:
    median = sorted_data[n//2]

# mode
counts = {}
for x in data:
    counts[x] = counts.get(x, 0) + 1
max_count = max(counts.values())
modes = [k for k, v in counts.items() if v == max_count]

print("Mean:", mean)
print("Median:", median)
print("Mode:", modes, "(every value appears once, so all are modes)" if max_count == 1 else "")
```

```python
#Confirming with NumPy and Pandas, plus Percentiles
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Variance:", np.var(data))        # population (matches Cell 15)
print("Std Dev:", round(np.std(data), 2))

print("25th percentile:", np.percentile(data, 25))
print("50th percentile:", np.percentile(data, 50))
print("75th percentile:", np.percentile(data, 75))

df.describe()
```

```python
# Correlation
corr = df["Python"].corr(df["Statistics"])
print("Correlation:", round(corr, 3))
```

5. Data Types in Analysis
Different data types need different handling.

```python
# DataFrame with every type
students = pd.DataFrame({
    "marks": [80, 75, 90, 60],                                   # numerical
    "department": ["DS", "CS", "DS", "IT"],                      # categorical
    "grade": ["A", "B", "A+", "C"],                              # ordinal
    "join_date": ["2023-06-15", "2023-06-20", "2023-07-01", "2023-07-05"],  # datetime
    "feedback": ["Great work", "Needs practice", "Excellent", "Average"]    # text
})

print(students.dtypes)
students
```

```python
# Handling each type
# Numerical: summary stats
print("Average marks:", students["marks"].mean())

# Categorical: count each category
print(students["department"].value_counts())

# Ordinal: set a real order so grades can be compared
grade_order = pd.CategoricalDtype(["C", "B", "A", "A+"], ordered=True)
students["grade"] = students["grade"].astype(grade_order)
print("Top grade:", students["grade"].max())

# Datetime: convert text to real dates, then extract parts
students["join_date"] = pd.to_datetime(students["join_date"])
students["join_month"] = students["join_date"].dt.month_name()
print(students[["join_date", "join_month"]])

# Text: simple cleaning
students["feedback_clean"] = students["feedback"].str.lower()
students["word_count"] = students["feedback"].str.split().str.len()
print(students[["feedback_clean", "word_count"]])
```

How each data type is handled:
- Numerical (marks): I use statistics like mean, median and standard deviation.
- Categorical(department): there is no natural order, so I count each category with `value_counts()`.
- Ordinal (grade): it has a real order (C < B < A < A+), so I set that order with `CategoricalDtype` and can then compare grades.
- Datetime (join_date): it is stored as text at first, so I convert it with `pd.to_datetime()` and extract parts like the month.
- Text (feedback): I clean it first by converting to lowercase, then measure it, for example with a word count.

6. Best Practices

```python
# Clean code example
# Bad: unclear names, no comments
x = [80, 75, 90]
y = sum(x) / len(x)

# Good: meaningful names, a docstring, and comments
def calculate_average(scores):
    """Return the average of a list of scores."""
    return sum(scores) / len(scores)

python_scores = [80, 75, 90]
average_score = calculate_average(python_scores)
print("Average score:", average_score)
```

```python
# Data handling and reproducibility
# Work on a copy so the original data stays safe
df_copy = df.copy()

# Check data quality before analysis
print("Missing values:\n", df_copy.isnull().sum())
print("Duplicate rows:", df_copy.duplicated().sum())

# Reproducibility: fixed random seed gives the same result every run
np.random.seed(42)
sample = np.random.choice(python_marks, size=3, replace=False)
print("Random sample:", sample)
```

Best practices I followed:
- Used meaningful variable names and docstrings.
- Checked for missing values and duplicates before analysis.
- Worked on a copy of the data to keep the original safe.
- Set a random seed so the results are reproducible.
- Interpreted results carefully: the correlation between Python and Statistics marks is high, but correlation does not mean causation. With only 10 students, the result may not apply to everyone.
